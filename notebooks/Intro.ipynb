```python
UPPER = "ABCÇDEFGĞHIİJKLMNOÖPRSŞTUÜVYZ"
LOWER = "abcçdefgğhıijklmnoöprsştuüvyz"
N = len(UPPER)


def caesar_encrypt(text, key):
    result = ""
    key = key % N

    for character in text:
        if character in UPPER:
            index = UPPER.index(character)
            result += UPPER[(index + key) % N]
        elif character in LOWER:
            index = LOWER.index(character)
            result += LOWER[(index + key) % N]
        else:
            result += character

    return result


def caesar_decrypt(ciphertext, key):
    return caesar_encrypt(ciphertext, -key)
```

```python
print("Harf sayısı:", N)
print("A:", UPPER.index("A"))
print("Ç:", UPPER.index("Ç"))
print("I:", UPPER.index("I"), "  ı:", LOWER.index("ı"))
print("İ:", UPPER.index("İ"), "  i:", LOWER.index("i"))
print("Ş:", UPPER.index("Ş"))
print("Z:", UPPER.index("Z"))
```

```python
text = "Anahtar yalnızca yetkili alıcıyla paylaşılacaktır."
key = 3

ciphertext = caesar_encrypt(text, key)

print("Plaintext :", text)
print("Key       :", key % N)
print("Ciphertext:", ciphertext)
```

```python
decrypted_text = caesar_decrypt(ciphertext, key)

print("Ciphertext    :", ciphertext)
print("Decrypted text:", decrypted_text)
```

```python
def show_caesar_steps(text, key):
    key = key % N
    print("Karakter\tİndeks\tİşlem\tYeni indeks\tSonuç")

    for character in text:
        if character in UPPER:
            alphabet = UPPER
        elif character in LOWER:
            alphabet = LOWER
        else:
            continue

        alphabet_index = alphabet.index(character)
        encrypted_index = (alphabet_index + key) % N
        encrypted_character = alphabet[encrypted_index]
        operation = f"({alphabet_index} + {key}) % {N}"
        print(
            f"{character}\t{alphabet_index}\t{operation}\t{encrypted_index}\t{encrypted_character}"
        )


show_caesar_steps(text, key)
```

```python
def visualize_caesar_alphabet(key):
    key = key % N
    cipher = "".join(UPPER[(i + key) % N] for i in range(N))
    print("Açık :", " ".join(UPPER))
    print("Şifre:", " ".join(cipher))


visualize_caesar_alphabet(key)
```
