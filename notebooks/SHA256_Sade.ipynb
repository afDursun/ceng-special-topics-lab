# SHA-256 — sade örnek
Bu notebook, `merhaba` metninin özetini ve küçük bir değişikliğin etkisini gösterir.

```python
import hashlib

metin = "merhaba"
ozet = hashlib.sha256(metin.encode("utf-8")).hexdigest()
print("Metin   :", metin)
print("SHA-256 :", ozet)
```

## Metin bilgisayarda baytlara dönüşür

```python
baytlar = metin.encode("utf-8")
print("Baytlar :", list(baytlar))
print("Hex     :", baytlar.hex(" "))
```

## Küçük değişiklik, farklı özet

```python
for deger in ["merhaba", "Merhaba", "merhaba!"]:
    ozet = hashlib.sha256(deger.encode("utf-8")).hexdigest()
    print(deger, "→", ozet)
```
